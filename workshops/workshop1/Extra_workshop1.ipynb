Eksempel A: tell tall innenfor et intervall

Oppgave: Lag count_between(values, low, high) som teller hvor mange tall som ligger mellom low og high (begge med). Bruk løkke.

```python
def count_between(values, low, high):
    count = 0
    for value in values:
        if low <= value <= high:
            count += 1
    return count

print(count_between([3, 8, 1, 10, 5], 3, 8))
print(count_between([1, 6], 0, 7))
```

Eksempel B: finn den varmeste dagen

Oppgave: Finn hvilken dag det var varmest, og temperaturen. Uten løkke.

```python
import numpy as np

days = np.array(['Mon', 'Tue', 'Wed', 'Thu', 'Fri' ])
temp = np.array([12.5, 14.5, 17, 17.3, 15.2])

i_max = np.argmax(temp)
temp_max = temp[i_max]
day_max = days[i_max]

print(f'Warmest: {temp_max} degrees on {day_max}')
```

Eksempel C: maksimer inntekt

Oppgave: Inntekt $R(p) = p \cdot (100 - 2p)$, der p er prisen. Finn prisen som gir høyest inntekt blant 101 kandidater mellom 0 og 50, vektorisert.

```python
import numpy as np

def revenue(p):
    return p * (100 - 2 * p)

prices = np.linspace(0, 50, 101)
r_all = revenue(prices)
i_max = np.argmax(r_all)
r_max = r_all[i_max]
p_max = prices[i_max]

print(f'Max revenue: {r_max} at price {p_max}')
```

Oppgave A: funksjon med løkke (15 min)

Lag en funksjon count_above(values, limit) som returnerer hvor mange tall i values som er større enn limit. Bruk en løkke, ikke NumPy.

```python
def count_above(values, limit):
    count = 0
    for value in values:
        if value > limit:
            count += 1
    return count

print(count_above([10, 2, 5, 7], 6))
print(count_above([1, 2], 10)) 
```

Oppgave B: finn den billigste (10 min)

En butikk selger en vare til ulike priser i forskjellige uker. Finn hvilken uke prisen var lavest, og hva prisen var. Gjør det uten løkke.

```python
weeks = np.array([1, 2, 3, 4, 5, 6, 7])
prices = np.array([112, 110, 120, 117, 105, 103, 123])

i_min = np.argmin(prices)
min_price = prices[i_min]
min_week = weeks[i_min]

print(f'Lowest price: {min_price} in week {min_week}')
```
