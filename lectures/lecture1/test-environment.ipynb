# Print information about the environment

- This notebook prints information about your computer, the Python version, and the installed packages.
- Run this notebook to test whether the central packages required in this course are installed in your Anaconda environment.
- For each package, the code either prints the installed version, or an error message if the package was not found.

```python
from IPython.display import Markdown, display
```

***
## Information about this computer

```python
# Print some information about this computer
import platform

display(
    Markdown(
        f"<span style='font-size: 18px;'><b>Operating System:</b> {platform.system()} {platform.release()} ({platform.version()})</span>"
    )
)
display(
    Markdown(
        f"<span style='font-size: 18px;'><b>Computer name:</b> {platform.node()}</span>"
    )
)
```

***
## Python version

```python
expected_version = '3.14.7'
py_version = platform.python_version()
if tuple(py_version.split('.')) == tuple(expected_version.split('.')):
    display(
        Markdown(
            f"<span style='font-size: 18px; color: green;'>✓ Python {py_version} is installed</span>"
        )
    )
else:
    display(
        Markdown(
            f"<span style='font-size: 18px; color: goldenrod;'>⚠ Python {py_version} is installed (expected {expected_version})</span>"
        )
    )
```

***
## Installed packages

```python
# Template strings for package availability messages
success_message = "<span style='font-size: 18px; color: green;'>✓ {package} {version} is installed</span>"
warning_message = "<span style='font-size: 18px; color: goldenrod;'>⚠ {package} {version} is installed (expected {expected})</span>"
failure_message = (
    "<span style='font-size: 18px; color: red;'>✗ {package} is not installed</span>"
)


def check_package(name, installed, expected):
    if tuple(installed.split('.')) == tuple(expected.split('.')):
        display(Markdown(success_message.format(package=name, version=installed)))
    else:
        display(
            Markdown(
                warning_message.format(
                    package=name,
                    version=installed,
                    expected=expected,
                )
            )
        )
```

```python
try:
    import numpy

    check_package('numpy', numpy.__version__, '2.5.2')
except ImportError:
    display(Markdown(failure_message.format(package='numpy')))
```

```python
try:
    import pandas

    check_package('pandas', pandas.__version__, '3.0.5')
except ImportError:
    display(Markdown(failure_message.format(package='pandas')))
```

```python
try:
    import matplotlib

    check_package('matplotlib', matplotlib.__version__, '3.11.1')
except ImportError:
    display(Markdown(failure_message.format(package='matplotlib')))
```

```python
try:
    import scipy

    check_package('scipy', scipy.__version__, '1.18.0')
except ImportError:
    display(Markdown(failure_message.format(package='scipy')))
```

```python
try:
    import openpyxl

    check_package('openpyxl', openpyxl.__version__, '3.1.5')
except ImportError:
    display(Markdown(failure_message.format(package='openpyxl')))
```
